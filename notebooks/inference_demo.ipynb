# AMPR Inference Demo

Predict Gene Ontology (GO) terms for your own proteins with the trained AMPR model.

**You need, per protein:** a sequence (FASTA) and a structure (PDB file at `pdbs/<id>.pdb`; AlphaFold models work).
ESM-2 3B runs best on a GPU (e.g. Colab T4/A100).

## 1. Setup

```python
# Clone the repo and install dependencies (skip the clone if already local)
# !git clone https://github.com/hungithust/datn_protein_function ampr
# %cd ampr
!pip install -q torch transformers h5py biopython pyyaml numpy
```

## 2. Get a trained checkpoint

Download `best.pt` for the branch you want into `checkpoints/<branch>_v3_esm3b/` (see the Releases / Google Drive link in the README). Example for MF:

```python
import os
os.makedirs('checkpoints/mf_v3_esm3b', exist_ok=True)
# e.g.  !gdown <FILE_ID> -O checkpoints/mf_v3_esm3b/best.pt
assert os.path.exists('checkpoints/mf_v3_esm3b/best.pt'), 'Put best.pt here first'
```

## 3. Prepare inputs

A FASTA of sequences and one PDB per protein, named `<id>.pdb` matching the FASTA header id.

```python
# Example: a single protein with id 'demo' (replace with your own)
with open('proteins.fasta', 'w') as f:
    f.write('>demo\nMKTAYIAKQRQISFVKSHFSRQLEERLGLIEVQAPILSRVGDGTQDNLSGAEKAVQVKVKALPDAQFEVVHSLAKWKR\n')
os.makedirs('pdbs', exist_ok=True)
# Place pdbs/demo.pdb (e.g. an AlphaFold model) before running the next cell.
```

## 4. Run prediction

```python
!python scripts/predict.py --branch mf \
  --fasta proteins.fasta --pdb_dir pdbs/ \
  --out predictions_mf.tsv --threshold 0.3
```
